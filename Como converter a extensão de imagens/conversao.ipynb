```python
from PIL import Image

import os
```

```python
caminho = "imagens/estrelas.png"

pasta_destino = ("imagensconvertidas")

imagem = Image.open(caminho)

imagem_rgb = imagem.convert("RGB")

imagem_rgb.save(f"{pasta_destino}/estrelas.jpg")
```

```python
imagem.save(f"{pasta_destino}/estrelas.tiff")
```

```python
lista_arquivos = os.listdir("imagens")

lista_arquivos
```

```python
extensao = input("Selecione o formato do arquivo")

for arquivo in lista_arquivos:
    
    imagem = Image.open(f"imagens/{arquivo}")
    
    
    imagem.save(f"{pasta_destino}/{arquivo}.{extensao}")
```
