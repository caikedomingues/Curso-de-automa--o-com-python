                            Pywhatkit

 Basicamente, ela serve como um "facilitador" para interagir com serviços
 web e ferramentas de sistema. Principais utilidades

 1. Automação de Whatsapp (A mais usada): Diferente da API oficial da Meta
 (que é paga e burocrática), o PyWhatkit usa o seu whatsapp web para enviar
 mensagens.

    -> Envio agendado: Você define o horário exato (hora e minuto) para
    a mensagem sair.

    -> Envio Instantâneo: Abre o navegador e envia na hora.

    -> Envio para grupos: Também permite automatizar mensagens para IDs
    de grupos.

    -> Envio de mensagens|: Pode enviar fotos com legendas automaticamente.

2. Integração com YouTube: Ela possui comandos simples para buscar e reproduzir
videos.

    -> Kit.playonyt("nome do video"): Esse comando abre o seu navegador padrão
    e ja começa a tocar o video mais relevante no youtube automaticamente. É
    ótimo para criar assistentes virtuais por voz.

3. Pesquisas no Google e Wikipédia: Você pode extrair informações da web
com uma única linha de codigo:

    -> Google Search: Abre o navegador com os resultados de uma busca.

    -> Info da Wikipédia: Resume um assunto e imprime no seu terminal, sem 
    você precisar abrir o navegador e ler páginas e páginas.


```python
# Observação: Para a biblioteca funcionar corretamente, é necessário 
# que o seu whatsapp web esteja conectado no seu navegador padrão.

import pywhatkit as kit

# Biblioteca que possibilita que o python acesse o sistema de arquivos
# d windows 
import os

# Função da biblioteca dotenv que tem como objetivo carregar no arquivo
# variáveis de ambientes.
from dotenv import load_dotenv
```

```python
# Irá carregar as variáveis de ambiente
load_dotenv()

# Função da biblioteca os que tem como objetivo acessar o valor da variável
# de ambiente. 
telefone = os.getenv("TELEFONE")

# Mensagem que iremos enviar na automaação.
mensagem = "Bora automatizar com python"

# Função da biblioteca pywhatkit que tem como objetivo acessar o seu whatsapp
# web e enviar a mensagem definida na automação. A função recebe como argumento
# o telefone (com o "+55" e o DDD da cidade) e a mensagem que será enviada.
kit.sendwhatmsg_instantly(telefone, mensagem)
```

```python
# Função da biblioteca pywhatkit que tem como objetivo definir um horário para o 
# envio da mensagem. A função recebe como argumento, o telefone, a mensagem, a hora e o minuto que você quer enviar a mensagem.
kit.sendwhatmsg(telefone, mensagem, 15, 48)
```

```python
# Função que tem como objetivo acesssar videos do youtube no seu navegador padrão
# A função recebe como argumento o nome do video que você quer acessar. A biblioteca ira acessar o video mais relevante daquele titulo que você digitou
kit.playonyt("slide way")
```
