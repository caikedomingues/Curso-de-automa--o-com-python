Import das bibliotecas necessárias

```python
# Import da biblioteca os que tem como objetivo conectar o python
# ao sistema de arquivos do sistema operacional do seu computador.
import os

# Tem como objetivo imprimir listas de forma organizada no terminal.
import pprint
```

Listando os arquivos do diretório "mover"

```python
# lista_arquivos: Ira conter a lista de arquivos e pastas presentes na pasta "mover" do projeto.

# listdir: Função que tem como objetivo listar os arquivos e pastas presentes
# em um diretório. A função pode receber como argumento o caminho da pasta
# que queremos listar os arquivos. Observação: Caso você opte por não passar
# um caminho, a função irá listar todos os arquivos da pasta do seu projeto.
lista_arquivos = os.listdir("mover")

# Ira imprimir as pastas e arquivos do diretório "mover"
pprint.pprint(lista_arquivos)
```

Movendo os arquivos para as suas respectivas pastas

```python
# Como os arquivos serão movidos para uma outra pasta, eu resolvi criar uma
# estrutura try/except que evitara as mensagens de erro do console do python
# caso o código seja executado mais de uma vez

# try: Ira inspecionar o bloco de código  com o objetivo de capturar possiveis
# erros de execução do código.

# except filenotfounderror: Ira tratar erros relacionados a inexistência de
# arquivos na pasta.
try:
    
    # Ira percorrer a lista de arquivos listados pelo listdir
    for arquivo in lista_arquivos:
    
    # Como a função lista os nomes dos arquivos, o conteudo da lista
    # irá lidar com dados do tipo texto, logo, poderemos verificar
    # o nome da extensão com o objetivo de evitar que o python mova
    # arquivos que não devem ser removidos. 
    
    # Irá verificar se é um arquivo excel através da verificação do nome
    # da extensão
     if 'xlsx' in arquivo:
         
         # Se essa condição for verdadeira, vamos verificar os nomes 
         # dos arquivos com o objetivo de direciona-los de forma correta
         # para as pastas.
        if 'Jan' in arquivo:
            
            # Se essa condição for verdadeira, vamos mover os arquivos
            # para a pasta de janeiro.
            
            # rename: Função do os que tem como objetivo mover e renomear
            # arquivos de um local para o outro. Ela recebe como argumento
            # o caminho do arquivo que será modificado (com o nome antigo
            # ) e o caminho de destino da pasta (com o novo nome ou o nome antigo
            # caso você queira apenas mover o arquivo.)
            os.rename(f'mover/{arquivo}',f'mover/Jan/{arquivo}' ) 
            
        elif 'Fev' in arquivo:
            
            # Se essa condição for verdadeira, vamos mover os arquivos
            # para a pasta de fvereiro.
            os.rename(f'mover/{arquivo}', f'mover/Fev/{arquivo}')
        
        elif 'Mar' in arquivo:
            
            # Se essa condição for verdadeira, vamos mover os arquivos
            # para a pasta de março.
            os.rename(f'mover/{arquivo}', f'mover/Mar/{arquivo}')

except FileNotFoundError:
    
    # Mensagem que irá aparecer se o python não encontrar os arquivos.
    print("Não há arquivos com essa extensão")
    
    
    
```
