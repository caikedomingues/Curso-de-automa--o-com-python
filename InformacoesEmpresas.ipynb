Cnpj Ws: Api brasielira muito utilizada no desenvolvimento de software para
automatizar o desenvolvimento de software para automatizar a consukta de dados
cadastrais de empresas através do número do cnpj. Em vez de uma pessoa ter
que entrar manualmente no site da Receita Federal, digitar i CNPJ e resolver
aqueles Captchas cansativos para descobrir as informações de uma empresa, o
programador integra essa API no sistema para fazer essa busca e forma 100%
automatizada e em frações de segundo.

```python
# Biblioteca do python que tem como objetivo realizar requisições 
# e se comunicar com sites e APIs
import requests

# Tem como objetivo imprimir valores de forma mais organizada.
import pprint
```

```python
# Cnpj que iremos usar no exercicio
cnpj = "33592510000154"

# Contém o link de requisição a API. Ela contém o valor do cnpj 
# que será consultado (via requisição GET)
link = f"https://publica.cnpj.ws/cnpj/{cnpj}"

# Função da biblioteca requests que tem como objetivo realizar requisições
# GET (consulta de dados) a sites e APIs. A função retorna um objeto que
# contém os resultados e métodos da requisição
requisicao = requests.get(link)

# Ira printar a resposta (status) do servidor que indica se a requisição
# funcionou ou não
print(requisicao)
```

```python
# O método json do objeto retornado pelo get serve para converter 
# o dados em json. O método irá retornar um dicionário que contém outros 
# dicionários e listas.
pprint.pprint(requisicao.json())
```

```python
# Acessando valores especificos do dicionário de valores. Para acessar
# esses valores devemos passar a estrutura correta de dicionários
# até chegar no valor buscado (no nosso caso, a descrição da atividade
# principal da empresa).
atividade_principal = requisicao.json()['estabelecimento']['atividade_principal']['descricao']

print(atividade_principal)
```
