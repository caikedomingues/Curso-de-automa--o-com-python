Biblioteca necessaria: pip install reportlab

Bibliotecas

```python
# Import da biblioteca pandas que tem como objetivo acessar e manipular dados
# de datasets
import pandas as pd

# Import do módulo canvas do módulo pdfgen da biblioteca reportlab.
# Esse módulo possui a classe necesssária para a manipulação de pdfs.
from reportlab.pdfgen import canvas

# Import da tupla que contem as medidas de uma folha A4. Ela pertence
# ao modulo pagesizes da subpasta lib da biblioteca reportlab. 
from reportlab.lib.pagesizes import A4
```

Acessando os dados

```python
# Função da biblioteca pandas que tem como objetivo acessar e manipular dados
# de um dataset. A função irá retornar um objeto DataFrame com os dados do
# dataset carregados na memória.
tabela = pd.read_csv('horas_extras.csv')

# Irá apresentar os dados da tabela.
display(tabela)
```

Criando 1 relatório: Primeiro, vamos criar um único relatório para aprendermos
como utilizar as ferramentas

```python
# Ira conter o nome do funcionário
nome = "Ana Souza"

# Irá conter o departamento do funcionário
departamento = "Financeiro"

# Ira conter a quantidade de horas extras do funcionário
horas_extras = 12

# Irá conter o mes e o ano em que as horas extras foram realizadas.
mes_ref = "Março/2025"

# Instância da classe Canvas do módulo canvas que tem como objetivo criar
# os pdfs. Ela recebe em seu construtor o nome dos arquivos e a tupla com
# medidas de PDF.
documento_pdf = canvas.Canvas(f"Relatório de horas extras {nome}.pdf", A4)

# titulo: Escolhe a fonte, escreve o texto

# Escolha da fonte: Método da classe Canva que possibilita a escolha
# e o tamanho da fonte do texto.
documento_pdf.setFont("Helvetica-Bold", 18)

# Distância e posicionamento do texto
x = 100

y = 750

# Texto que será o titulo do documento
texto = f"Relatório de horas extras da {nome}"

# Método da classe Canva que tem como objetivo definir o texto do 
# documento. O método recebe como argumento o posicionamento do
# texto (x e y) e o texto do documento (texto)
documento_pdf.drawString(x, y, texto)

# escreve o texto: escolha da fonte, escreve o texto

# Escolha da fonte e do tamanho do texto.
documento_pdf.setFont("Helvetica", 12)

# Texto do relatório
documento_pdf.drawString(100, 700, f"Nome: {nome}")

documento_pdf.drawString(100, 670, f"Departamento: {departamento}")

documento_pdf.drawString(100, 640, f"horas extras: {horas_extras}")

documento_pdf.drawString(100, 610, f"Mês de Referência: {mes_ref}")

documento_pdf.drawString(100, 580, "Esse relatório foi gerado automaticamente com python")

# salvar o relatório

documento_pdf.save()
```

Criando um relatório para cada pessoa da base de dados: Agora a ideia é percorrer
a tabela de dados com o objetivo de criar um relatório para cada funcionário.

```python
# For que ira percorrer todas as linhas da tabela de dados.
# index: Atributo da classe DataFrame que contém os indices de cada
# linha da tabela
for linha in tabela.index:
    
    # loc: Atributo que acessa uma lista de valores com base em uma condição.
    # Nesse caso, ele irá receber a linha e a coluna que queremos acessar
    # na tabela.
    nome = tabela.loc[linha, "Nome"]

    departamento = tabela.loc[linha, "Departamento"]

    horas_extras = tabela.loc[linha, "Horas Extras"]

    mes_ref = tabela.loc[linha, "Referência"]

    # Instância da classe Canvas
    documento_pdf = canvas.Canvas(f"Relatório de horas extras {nome}.pdf", A4)

    documento_pdf.setFont("Helvetica-Bold", 18)

    x = 100

    y = 750

    texto = f"Relatório de horas extras da {nome}"

    documento_pdf.drawString(x, y, texto)

    # escreve o texto: escolha da fonte, escreve o texto
    documento_pdf.setFont("Helvetica", 12)

    documento_pdf.drawString(100, 700, f"Nome: {nome}")

    documento_pdf.drawString(100, 670, f"Departamento: {departamento}")

    documento_pdf.drawString(100, 640, f"horas extras: {horas_extras}")

    documento_pdf.drawString(100, 610, f"Mês de Referência: {mes_ref}")

    documento_pdf.drawString(100, 580, "Esse relatório foi gerado automaticamente com python")
    
    # corpo

    # salvar o relatório

    documento_pdf.save()
```
