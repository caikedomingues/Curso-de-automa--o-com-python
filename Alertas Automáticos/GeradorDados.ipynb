Função do arquivo: Esse arquivo irá gerar numeros (dados) aleatórios que 
serão armazenados no banco de dados firebase do google

link do site do firebase: firebase.google

bibliotecas necessárias para o exercicio

```python
# Import da biblioteca que possui a função randint que tem como objetivo
# gerar aleatoriamente numeros inteiros.
import random

# Biblioteca que possibilita a manipulação de datas.
import datetime

# Biblioteca que possui a manipulação do tempo 
import time

# Biblioteca que possibilita o envio de requisições a sistemas, APIs
# e sites. 
import requests
```

Gerador de dados aleatórios para o banco firebase

```python
# Loop que irá rodar o trecho de código infinitamente.
while True:
    
    # now: Métod da classe datetime da biblioteca datetime
    # que tem como objetivo pegar a data e a hora atual do
    # nosso sistema.
    data = datetime.datetime.now()
    
    # Função que tem como objetivo gerar valores aleatórios. A função
    # irá gerar valores aleatórios entre 100 e 200.
    vazao = random.randint(100, 200)
    
    # String formatada que irá conter os valores que serão enviados ao
    # banco de dados firebase (data e valor aleatório da vazão). ela deverá
    # conter a estrutura de um arquivo JSON.
    logs = f'{{"Data": "{data}", "Vazao": "{vazao}"}}' 
    
    # post: Método da biblioteca requests que tem como objetivo enviar
    # requisições POSTs (envio de dados) a sistemas (no nosso caso, ela
    # será responsável por enviar os dados ao firebase). A função vai
    # receber como argumento o link do servidor do banco e os dados que
    # devem ser enviados (presentes no argumento data da função post). 
    requisicao = requests.post("https://alarme-b3d19-default-rtdb.firebaseio.com/.json", data=logs)
    
    # Ira dar um intervalo de 10 segundos antes de reiniciar toda a execução
    # do bloco.
    time.sleep(10)
```
