Bibliotecas necessárias

```python
# Classe da biblioteca bs4 que tem como objetivo extrair informações de sites
# através do HTML.
from bs4 import BeautifulSoup

# Biblioteca que tem como objetivo realizar requisições a sites e APIs
import requests
```

Extraindo informações de sites

```python
# Irá conter o link do site que iremos extrair as informações.
url = "https://www.nationalgeographicbrasil.com/"

# Função que irá acessar o site e extrair informações.
def extrair_informacoes():
    
    # Como há diferença nas requisições feitas por um usuário e uma linguagem
    # de programação, precisamos "mascarar" a nossa requisição como se fossemos
    # um usuário, dessa forma, teremos a resposta que um usuário receberia.
    # Para isso, precisamos criar um dicionário que nomeado de "User-Agent"
    # que ira conter como valor a versão atual do nosso navegador. 
    headers = {"User-Agent": "Chrome/148.0.7778.179"}
    
    # get: Função da biblioteca requests que tem como objetivo realizar 
    # requisições GET (consultas de dados) a sites. A função recebe como
    # argumento a url e o dicionário definido anteriormente. Ela retorna
    # um objeto do tipo response que contém os dados retornados pelo site. 
    requisicao = requests.get(url, headers=headers)
    
    # Ira verificar se o status da nossa consulta é igual a 200 (servidor
    # ok) o que significa sucesso na requisição.
    # status_code: Atributo da classe response retornada pela função get
    # que contém o status do sercidor 
    if requisicao.status_code == 200:
        
        # Instância da classe BeautifulSoup que extrai informações
        # de páginas web através do HTML. A xlasse recebe em seu
        # construtor: o valor do atributo text da classe response
        # e o "html.parser" que é o comando que indica que o
        # BeautifulSoup deve acessar o HTML da página. 
        pagina = BeautifulSoup(requisicao.text, "html.parser")
        
        # Ira conter todos os titulos da página acessada. 
        # find_all: Método da classe BeautifulSoup que tem como objetivo buscar
        # elementos HTML da página. O método irá receber como argumento o comando
        # class_ que recebe o nome da classe que deve ser buscada. Nesse caso
        # eu coloquei a classe presente em todos os h2 da página.
        titulos = pagina.find_all(class_="css-8lpzv")
        
        # Ira percorrer a lista de valores retornado pelo find_all
        for titulo in titulos:
            # Ira imprimir os valores presentes no atributo text da BeautifulSoup
            print(titulo.text)
        
    
```

Chamada da função

```python
extrair_informacoes()
```
