Biblioteca Necessária

```python
# Biblioteca que possibilita a manipulação de planilhas
import openpyxl
```

Atualizando a planilha

```python
# Função que irá inserir valores na planilha.
def atualizar_planilha():
    
    # load_workbook: Função da biblioteca openpyxl que tem como objetivo
    # acessar o arquivo excel que queremos editar. A função recebe como
    # argumento o caminho do arquivo excel.A função irá retornar um
    # objeto workbook que irá representar o nosso arquivo e as suas
    # propriedades.
    arquivo = openpyxl.load_workbook('excel/dados.xlsx')
    
    # active: atributo da classe criada pela função load_workbook que serve para # 
    # especificar qual aba do excel (sheet ou página) queremos editar. 
    # No nosso caso, como o nosso arquivo só possui uma aba, não precisamos # 
    # especificar o nome da aba.
    
    aba = arquivo.active
    
    # Função que tem como objetivo inserir valores na planilha. Ela recebe
    # como argumento a lista de valores que irão compor as linhas da tabela
    # excel.
    aba.append(["Lira", 31, "programação"])
    
    # Ira salvar as alterações no caminho especificado (se for na mesma pasta,
    # o python ira apenas atualizar o arquivo ja existente no caminho)
    arquivo.save("excel/dados.xlsx")
    
    # mensagem de sucesso.
    print("Planilha atualizada")
    
    
    
```

Chamada da função

```python
atualizar_planilha()
```
