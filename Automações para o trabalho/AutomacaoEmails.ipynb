Bibliotecas Necessárias

```python
# Módulo do python que implementa o protocolo SMTP (Protocolo de Transferência
# de Correio Simples). Ele é o "carteiro" do nosso código. É este módulo que cria
# uma conexão segura com o servidor de e-mail (como o Gmail, Outlook ou Yahoo),
# faz a autenticação com o seu usuário e senha e gerencia a viagem dos dados
# pela internet até o destinatário.  
import smtplib 

# Serve para criar o envelope da mensagem. Um email não é apenas um bloco de texto
# puro, ele tem metadados anexados. Esse comando cria um objeto "container" vázio
# onde vamos colocar de forma organizada:
# -> Quem está enviando (From)
# -> Quem vai receber (To)
# -> O assunto do e-mail(subject)
# Além de permitir que o e-mail carregue multiplas partes juntas (como o texto
# do corpo e um arquivo anexo, como uma planilha de cotações ou uma imagem).
from email.mime.multipart import MIMEMultipart

# Serve para criar a folha de papel com o conteúdo que vai dentro do envelope.
# É com ele que vamos definir a mensagem real do e-mail. Ele possui um argumento
# muito importante chamado _subtype. Com ele, podemos definir como o nosso texto
# será lido como:
# -> "plain": Texto puro e simples, sem nenhuma formatação
# -> "html": Texto formatado com tags HTML. Isso permite o envio de e-mails
# elegantes com tabelas, textos em negrito, cores, links clicaveis ou imagens
# incorporadas.  
from email.mime.text import MIMEText

# função da biblioteca dotenv que tem como objetivo carregar variáveis de
# ambientes no nosso código
from dotenv import load_dotenv

# Biblioteca que possibilita que o python se conecte com o nosso sistema
# de arquivos.
import os
```

Carregando as variáveis de Ambiente

```python
# Ira carregar as variáveis de ambientes no nosso código.
load_dotenv()

# Ira acessar os valores das variáveis de ambiente (email e senha)
EMAIL = os.getenv('EMAIL')

SENHA = os.getenv('SENHA')
```

Criação do envio de emails

```python
# Função que irá enviar o e-mail para o destinatário.
def enviar_email():
    
    # Irá conter o e-mail presente na variável de ambiente EMAIL
    remetente = EMAIL
    
    # Ira conter a senha presente na variável de ambiente SENHA
    senha = SENHA
    
    # Irá conter o e-mail que receberá a mensagem (o mesmo e-mail do remetente)
    destinatario = EMAIL
    
    # assunto/titulo do e-mail.
    assunto = "Relatório do dia de hoje"
    
    # Texto/mensagem do e-mail
    corpo = "Este é um email automatico do relatório do dia de hoje"
    
    # Instância da classe MIMEMultipart que irá definir a mensagem do
    # e-mail.
    mensagem = MIMEMultipart()
    
    # Nessa etapa, iremos atribuir valores ao dicionário da classe. Esses valores
    # irão representar a estrutura do nosso e-mail.
    
    # Ira conter o remetente (quem esta enviando a mensagem)
    mensagem["From"] = remetente
    
    # Ira conter o destinatário (quem recebe a mensagem)
    mensagem["To"] = destinatario
    
    # Ira conter o titulo/assunto do e-mail
    mensagem["Subject"] = assunto
    
    # Define que o nosso e-mail sera texto bruto, ou seja, sem nenhum tipo
    # código HTML
    mensagem.attach(MIMEText(corpo, "plain"))
    
    # SMTP: Função que tem como objetivo definir o padrão de envio de e-mails.
    # Ela recebe como argumento o código de smtp da sua plataforma de e-mail
    # e a porta que o código irá operar o envio do email. Ela também é respon
    # sável pela conexão com o servidor de e-mails através do código smtp.
    servidor = smtplib.SMTP("smtp.gmail.com", 587)
    
    # Ira criptografar o envio do nosso e-mail.
    servidor.starttls()
    
    # Função da smtplib que irá logar no email do remetente usando o email
    # e a senha definida pelo usuário.
    servidor.login(remetente, senha)
    
    # Irá enviar o email usando o remetente, o destinatario e a mensagem.
    # as_string(): Método da classe MIMEMultpart que define que o nosso
    # e-mail sera uma string (dados do tipo texto)
    servidor.sendmail(remetente, destinatario, mensagem.as_string())
    
    # Ira encerrar a conexão com o servidor de e-mails
    servidor.quit()
    
    # Impressão da mensagem de sucesso do código
    print(f"Email enviado com sucesso")
    
```

Chamada da função de envio de e-mails

```python
enviar_email()
```
