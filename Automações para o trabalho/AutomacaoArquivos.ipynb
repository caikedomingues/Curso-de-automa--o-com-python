Biblioteca necessária

```python
# Biblioteca que possibilita que o python manipule o sistema de arquivos do
# sistema operacional.
import os
```

Organuzando os arquivos


```python
# Função que ira adicionar os arquivos em uma pasta.
def organizar_arquivos():
    
    # Variável que irá conter o caminho da pasta que queremos inserir os arquivos
    pasta = "arquivos"

    # Ira verificar se o diretório especificado na variável pasta existe no
    # sistema.
    # exists: Função do módulo path da biblioteca os que verifica se uma
    # pasta ou arquivo existe no sistema de arquivos. A função recebe como 
    # argumento o caminho da pasta ou arquivo que queremos verificar. Ela
    # retorna um valor booleano que indica o resultado da função.
    if  not os.path.exists(pasta):
    
      # Se essa condição for verdadeira, ou seja, se a pasta não existir,
      # vamos usar a função makedirs da os para criar o diretório. A função
      # recebe como argumento o caminho que irá conter a pasta criada
      os.makedirs(pasta)

    # Ira listar os arquivos da pasta do nosso projeto. Por padrão, quando
    # não definimos argumentos na função, ela lista todos os arquivos e 
    # diretórios da pasta do nosso projeto python.
    lista_arquivos = os.listdir()
    
    # Ira percorrer a lista de arquivos do projeto com o objetivo de acessar
    # todos os arquivos e pasta do diretório.
    for arquivo in lista_arquivos:
        
        # Ira verificar se a extensão .txt existe no nome do arquivo (estamos
        # lidando com dados do tipo texto).
        if ".txt" in arquivo:
            
            # Se essa condição for verdadeira, vamos usar a função rename
            # do os que tem como objetivo renomear e transferir arquivos
            # e pastas. A função recebe como argumento o caminho atual
            # (com o nome atual) e o caminho de destino (com o novo nome
            # ou o nome antigo, caso voê queira apenas transferir os arquivos)
            os.rename(arquivo, f"{pasta}/{arquivo}")
    
    # Impressão da mensagem de sucesso
    print("Arquivos organizados com sucesso")
```

Chamada da função

```python
organizar_arquivos()
```
