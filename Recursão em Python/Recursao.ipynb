                                     Recursão
-> Em programação, a recursão acontece quando uma função chama a si mesma dentro
do nosso próprio código para resolver um problema. 

-> Devemos pensar nela como uma boneca russa (Matrioshka): para chegar na menor
boneca de todas, você prcisa abrir uma boneca maior, que contém outra menor, 
que contém outra menor, e assim por diante.

-> Para que uma função recursiva não rode infinitamente e trave o seu computador
(o erro RecursionError: maximum recursion depth exceeded), ela precisa 
obrigatoriamente de duas coisas:

    1. Caso Base (Condição de Parada): É a condição que faz a função parar de
    chamar a si mesma. É o momento em que chegamos na menor boneca.

    2. Caso Recursivo: É a parte do código onde a função quebra o problema
    em uma parte menor e chama a si mesma novamente.


                Como o Python processa isso por trás dos panos?
-> O python gerencia as funções recursivas usando uma estrutura de dados
na memória chamada Pilha de Execução (Call Stack). Ele vai "empilhando" as
chamadas de função uma em cima da outra e só começa a calcular os valores
de volta quando encosta no Caso Base.

                             Quando usar recursão?

-> A recursão é excelente e deixa o código muito elegante quando lidamos com
estruturas de dados que são naturalmente ramificadas ou repetitivas como:

    -> Navegat por estruturas de pasta e subpastas no sistema operácional.

    -> Percorrer árvores de dados e objetos JSON complexos cheios de subchaves.

    -> Algoritmos de busca e ordenação complexos (como Quick Sort ou Merge Sort).

-> Conseguiu pegar a lógica de como a função vai "se empilhando" na memória? Se

Desafio da Aula: Backup do sistema de Vendas. Vamos criar uma função recursiva
que ira listar todos os arquivos de pastas e subpastas. Dessa forma, poderemos
verificar quais arquivos nós ja temos e quais precisarão de backup

```python
# lista de meses do sistema. essa lista contém os meses que possuem arquivos de
# vendas.
meses_sistema = ["jan2022","fev2022","mar2022","abr2022","mai2022","jun2022","jul2022","ago2022","set2022","out2022","nov2022","dez2022","jan2021","fev2021","mar2021","abr2021","mai2021","jun2021","jul2021","ago2021","set2021","out2021","nov2021","dez2021","jan2020","fev2020","mar2020","abr2020","mai2020","jun2020","jul2020","ago2020","set2020","out2020","nov2020","dez2020","jan2019","fev2019","mar2019","abr2019","mai2019","jun2019","jul2019","ago2019","set2019","out2019","nov2019","dez2019","jan2018","fev2018","mar2018","abr2018","mai2018","jun2018","jul2018","ago2018","set2018","out2018","nov2018","dez2018","jan2017","fev2017","mar2017","abr2017","mai2017","jun2017","jul2017","ago2017","set2017","out2017","nov2017","dez2017"]

# Ira imprimir o tamanho da lista no sistema.
print(len(meses_sistema))
```

Construindo a função de recursão

```python
# Biblioteca que tem como objetivo possibilitar que o python se conecte ao
# sistema de arquivos do sistema operacional.
import os

# Biblioteca que possibilita a impressão de dados e listas de forma mais 
# organizada.
import pprint

# Lista que ira conter todos os arquivos extraidos das pastas e subpastas.
arquivos_extraidos = []

# Função que irá listar os arquivos e adiciona-los na lista de arquivos extraidos.
# A função recebe como argumento, o path (caminho) da pasta que você quer extrair
# os arquivos.
def pegar_arquivos_pasta(pasta):
    
    # Função da biblioteca os que tem como objetivo listar os arquivos de uma
    # pasta. A função recebe como argumento o caminho da pasta que queremos
    # listar.
    arquivos = os.listdir(pasta)
    
    # For que irá percorrer a lista de arquivos da pasta escolhida.
    for arquivo in arquivos:
        
        # Como a lista de arquivos possuem dados do tipo texto, vamos verificar
        # se o nome do arquivo possui o ".txt" e o "vendas". Dessa forma teremos
        # certeza de duas coisas:
        
        # 1 -> que estamos coletando um arquivo, pois pastas não possuem extensões
        
        # 2 -> que vamos coletar apenas arquivos relacionados a vendas.
         if ".txt" in arquivo and "Vendas" in arquivo:
             
             # Se essa condição for verdadeira, vamos iniciar o processo de inserção de arquivos
             # na lista.
             
             # Nessa etapa vamos dividir o nome do arquivo, acessar a ultima parte do nome (extensão do
             # arquivo) e substitui-lo por um espaço em branco. Dessa forma, poderemos inserir os nomes de
             # forma mais limpa e organizada.
             # split: Função do python que separa as palavras de uma string
             # [-1]: Acessa a ultima da divisão da string (extensão do arquivo)
             # replace: Tem como objetivo substituir palavras em strings
             
             nome_mes = arquivo.split()[-1].replace(".txt", "")
            
            # Ira inserir o nome do arquivo na lista de arquivos extraidos.
             arquivos_extraidos.append(nome_mes)
        
        # Se o nome do arquivo não tiver o ".txt" (o que significa que não há a existência de uma extensão),
        # significa que estamos lidando com uma pasta
         elif ".txt" not in arquivo:
            
            # Se essa condição for verdadeira, vamos chamar a própria a função dentro dela mesma
            # (recursão) com o objetivo de acessar os arquivos da subpasta que está presente na 
            # pasta atual. Basicamente ela vai fazer o mesmo processo, porém com as subpastas da
            # pasta principal.
            pegar_arquivos_pasta(f'{pasta}/{arquivo}')
            
        # Resumindo: O loop ira ficar percorrendo a pasta passada como argumento com o objetivo de acessar todos os arquivos e subpastas
        # da pasta principal (pasta passada como argumento)
    
    # Ira imprimir a lista de arquivos retornadas pelo listdir.
    pprint.pprint(arquivos)
    
```

Chamada da função

```python
# Vamos chamar a função de pegar arquivos para visualizar a lista de arquivos e extrair os arquivos de vendas.
# Observação: Nem todos os arquivos que apareceram na lista irão ser extraidos, pois a função coleta apenas
# os arquivos de vendas.
pegar_arquivos_pasta('Arquivos')
```

Verificando quais meses não estão na lista de arquivos extraidos


```python
# Ira percorrer a lista de meses  
for mes in meses_sistema:
    
    # Ira verificar se o nome do mes está presente nos nomes dos arquivos da lista de extraidos
    # com o objetivo de coletar os meses que precisam de backup dos seus arquivos.
    if mes not in arquivos_extraidos:
        
        # Impressão do meses que não estão presentes na lista de extraidos.
        print(mes)
        
```
