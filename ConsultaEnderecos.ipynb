```python
# Biblioteca do python que possibilita requisições a sites e APIs   
import requests
```

Consulta de Informações de um CEP

```python
# Irá conter o cep que queremos pesquisar na API
cep = "20090002"

# Link que irá conter a informação que iremos requisitar a API.
# O link contém o numero do cep e o formato que a api ira retornar
# os dados.
link = f'https://viacep.com.br/ws/{cep}/json/'

# Ira utilizar a função get para enviar uma requisição get a API
# para buscar informações. A função ira ter como retorno o status
# do servidor, ou seja, se a sua requisição deu certo ou não.
# Segundo a documentação há 2 possibilidades de retorno: 
# 200 (ok) para requisições que estão corretas
# 400 (bad request)  para requisições que não funcionaram por conta de erros
# de sintaxe na escrita do cep.
requisicao = requests.get(link)

# Ira imprimir o status da api
print(requisicao)

# Ira imprimir os dados em formato json que contém um dicionário 
# que atribui rótulos aos dados.
print(requisicao.json())

# Ira conter o dicionário json com o objetivo de filtrar os dados
# através de rótulos.
dicionario_requisicao = requisicao.json()

# Filtrando valores do dicionário através dos rótulos.
uf = dicionario_requisicao['uf']

cidade = dicionario_requisicao['localidade']

bairro = dicionario_requisicao['bairro']

# Impressão dos valores filtrados
print("UF: ", uf)

print("CIDADE: ", cidade)

print("BAIRRO", bairro)
```

Busca de CEP a partir de um endereço

```python
# Nessa etapa será necessário definir 3 argumentos para o link da requisição a API: o estado, a cidade e o endereco
estado = "RJ"

cidade ="Rio de Janeiro"

endereco = "Rio Branco"
 
# Link que ira conter as informações que serão requisitadas a API
link = f'https://viacep.com.br/ws/{estado}/{cidade}/{endereco}/json/'

# Função do python que irá realizar uma requisição (do tipo GET) a API
requisicao = requests.get(link)

# Ira imprimir o status da API que irá indicar se a requisição funcionou ou não.
print(requisicao)

# Ira retornar uma lista de dicionários com os ceps das regiões que possuem os valores próximos dos passados ao link da API
print(requisicao.json())
```
