                                yfinance
-> É uma biblioteca de código aberto (open-source) para python desenvolvida
com um propósito muito claro: buscar dados financeiros e cotações de mercado
diretamente do Yahoo Finance de forma gratuita, rápida e estruturada. 

-> No universp da Análise de Dados e Finanças Quantitativas, ela se tornou
a ferramenta mais popular para substituir APIs pagas ou bibliotecas legadas
(como o antigo pandas_datareader).

                            O que ela faz na prática?
-> O grande trunfo da yfinance é que ela conversa com os servidores do Yahoo,
coleta dados brutos da internet e os converte automaticamente em um pandas.
DataFrame. Isso significa que, com apenas uma linha de código, você ganha
uma tabela perfeitamente limpa e pronta para análise. Ela permite que
você extraia informações como:

    Preços históricos de ações (abertura, fechamento, máximas, mínimas e volume
    de negociação).

    Dados de ativos globais e nacionais (como ações da B3, criptomoedas, fundos
    imobiliarios e indices como o ibovespa ^BVSP ou S&P500 ^GSPC)

    Informações fundamentalistas (balanço patrimonial, demonstrativo de
    resultados, dividendos distribuidos e desdobramentos de ações).

                    Como ela se organiza? (os principais componentes)

-> Para trabalhar com ela, você basicamente usará dois caminhos principais:

    1 - O método rápido em bloco (yfinance.download): Ideal para quando queremos só puxar o histórico de preços de um ou mais ativos de uma vez para montar
    um gráfico ou calcular rentabilidade.

    2. O objeto ticker (yfinance.Ticker): Ideal para quando precisamos fazer
    uma análise profunda da empresa especifica. Você cria um "objeto cliente"
    para aquela ação e ganha acesso a vários metodos internos.

                            Por que ela é tão usada?
-> Gratuita e sem necessidade de chave de API: Muitas plataformas de dados
financeiros exigem que você crie uma conta, cadastre cartão e use tokens/
chaves de acesso limitadas. A yfinance funciona direto, sem burocracia.

-> Integração nativa com ecossistema de dados: Como o retorno dela é um
DataFrame do pandas, você pode plugar os dados imediatamente no Matplotlib
ou Seaborn para plotar gráficos de linha das cotações, ou no Scikit-Learn
se estiver montando modelos de regressão para tentar prever tendências.


```python
# Import da biblioteca yfinance.
import yfinance
```

Visualizando cotações do ibovespa

```python
# Lista que irá conter o código das empresas que queremos acessar no Yahoo
# Finance
ativos = ["ITUB3.SA", "VALE3.SA", "PETR3.SA", "^BVSP"]

# Ira acessar o valor do indice 3 (empresa de indice 3)
ticker = ativos[3]

# Data inicial e data_final: Intervalo de datas que iremos acessar para coletar
# os dados.
data_inicial = "2026-01-01"

data_final = "2026-04-30"

# tabela_cotações: Ira conter as cotações extraidas da API da Yahoo

# download: Função da biblioteca yfinance que realiza as requisições
# a API e coleta os dados das cotações da empresa. A função recebe 
# como argumento:
# ticker: Código da empresa na API do yahoo finance.
# start: Recebe a data inicial do intervalo que queremos acessar.
# end: Recebe a data final do intervalo que queremos acessar
# Observação: A função retorna um Dataframe.
tabela_cotacoes = yfinance.download(ticker, start = data_inicial, end = data_final)

# Acessando as 10 primeiras linhas do Dataframe retornado
tabela_cotacoes.head(10)
```

Verificando as principais informações do dataframe utilizando a função info

```python
tabela_cotacoes.info()
```

Filtrando colunas

```python
# É possivel filtrar valores do dataframe através do nome das colunas.
close = tabela_cotacoes['Close']

# Ira mostrar os valores filtrados do DataFrame.
display(close)
```

Visualizando todas as cotações

```python
# Nesse caso, ao invés de passar um ativo especifico, vamos passar a lista
# inteira de ativos. 
todos_ativos = yfinance.download(ativos, start=data_inicial, end=data_final)

# Ira mostrar os valores do dataframe
display(todos_ativos)
```

```python
# Também é possivel filtrar as colunas logo após a requisição, basta filtrar
# os dados assim que a função download realizar a requisição a API. Dessa forma,
# a variável todos_ativos irá mostrar apenas a coluna filtrada
todos_ativos = yfinance.download(ativos, start=data_inicial, end=data_final)['Close']

# Impressão dos valores filtrados na requisição
display(todos_ativos)
```
