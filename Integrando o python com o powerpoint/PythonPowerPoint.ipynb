Criar PPT

```python
# import da biblioteca ptx que possui a classe Presentation que possui os métodos e atributos de manipulação do powerpoint 
from pptx import Presentation

# Instância da classe Presentation que criará a apresentação do powerpoint
# (o objeto apresentação)
apresentacao = Presentation()
```

Criar PPT baseado em Modelo: Nessa etapa, vamos aprender como acessar os layouts pré definidos pelo powerpoint

```python

# editar o ppt: Sempre devemos seguir a ordem: adicionar elemento -> editar o elemento

# Esse trecho irá criar os slides das nossas apresentações:
# apresentação: Classe que representa a apresentação do powerpoint

# slides: Atributo que contém as funções de edição de slides. Isso
# é possivel através de uma técnicia chamada de composição, que consiste
# na criação de uma classe que contém métodos que serão atribuidos em
# atributos de uma outra classe (a classe que sera utilizada na 
# criação de uma tarefa ou projeto). Dito isso, a classe Presentation
# utiliza uma outra classe que atribui ao atributo slides, funções necessárias
# para a edição e criação de slides.

# add_slides: Função do atributo slides que cria slides na apresentação. A
# função recebe como argumento o tipo do layout que você quer utilizar.

# slides_layouts: Atributo da classe Presentation que possui a lista de
# layouts pré definidos pelo powerpoint.
slide1 = apresentacao.slides.add_slide(apresentacao.slide_layouts[0])
slide2 = apresentacao.slides.add_slide(apresentacao.slide_layouts[1])
slide3 = apresentacao.slides.add_slide(apresentacao.slide_layouts[2])
slide4 = apresentacao.slides.add_slide(apresentacao.slide_layouts[3])
slide5 = apresentacao.slides.add_slide(apresentacao.slide_layouts[4])
slide6 = apresentacao.slides.add_slide(apresentacao.slide_layouts[5])
slide7 = apresentacao.slides.add_slide(apresentacao.slide_layouts[6]) #slide em branco
slide8 = apresentacao.slides.add_slide(apresentacao.slide_layouts[7])

# placeholders: servem para acessar os elementos do slides através de indices.
# titulo: Irá acessar o campo de titulo do slide
# subtitulo: Irá acessar o campo de subtitulo do slide.  
titulo = slide1.placeholders[0]

subtitulo = slide1.placeholders[1]

# text: o text é um atributo da classe da lista placeholder que
# recebe o conteudo de texto do componente que você quer acessar.
titulo.text = "1° slide da apresentação"

subtitulo.text = "Estamos criando ppt com Python"

# salvar esse ppt: Ira criar o arquivo pptx e salvará todas as alterações 
# feitas na apresentação
apresentacao.save("MeuPPT.pptx")
```

Criar PPT personalizados: Vamos aprender a criar um slide 

```python
# A classe Inches do módulo util da pptx serve como um conversor de medidas.
# Ela permite que você defina tamanhos usando polegadas (Inches), e ela
# mesma faz o cálculo matemático pesado por tras dos panos para converter
# isso na unidade que o PowerPoint entende.
from pptx.util import Inches

# A classe pt do módulo util da pptx serve para definir o tamanho da
# fonte de textos no powerpoint
from pptx.util import Pt

# Nessa etapa, vamos escolher o slide de indice 6 que é o slide em branco
slide = apresentacao.slides.add_slide(apresentacao.slide_layouts[6])

# Editar o ppt
# x e y: Definem a posição da caixa de texto onde x representa a distância
# horizontal e y a distância vertical. Esses valores são definidos através
# da classe inches que recebe em seu construtor o valor de cada distância
x = Inches(1)
y = Inches(1)
# Define a largura e a altura da caixa de texto
largura = Inches(2)
altura = Inches(2)

# caixa_texto: Ira conter a caixa de texto que será manipulada na
# apresentação. 

# shapes: Atributo que possui métodos que criam elementos visuais
# nos slides da apresentação.

# add_textbox: método do atributo shapes que cria caixas de texto nas
# apresentações de slides. A função recebe como argumento a distância
# horizontal e vertical (x e y), a largura (largura) e a altura (altura)  
caixa_texto = slide.shapes.add_textbox(x, y, largura, altura)

# text: Atributo que contém o conteudo do texto da caixa de texto
caixa_texto.text = "Slide de vendas do mes de janeiro"

# Atributo que contém métodos de manipulação de textos nas apresentações   
text_frame = caixa_texto.text_frame

# add_paragraph: Função do atribuo text_frame que cria paragrafos em texto
paragrafo = text_frame.add_paragraph() 

# Ira conter o texto do paragrafo
paragrafo.text = "R$10.000"

# Ira criar um novo paragrafo.
paragrafo = text_frame.add_paragraph()

# Ira criar o texto do novo paragrafo
paragrafo.text = "Produto mais vendido: iphone"

# bold: atributo do objeto font que recebe um valor booleano indicando se as fonte
# do paragrafo/texto deve ficar ou não em negrito 
paragrafo.font.bold = True

# size: Atributo do objeto que define o tamanho da fonte do texto
# Pt: classe que recebe em seu construtor o tamanho da fonte.  
paragrafo.font.size = Pt(30)


# Salva o ppt
apresentacao.save("MeuPPT2.pptx")
```

Criar PPTs com Gráficos

```python
from pptx.util import Inches

# O categoryChartData é uma classe do módulo data da biblioteca pptx que tem como objetivo inserir dados e categorias no grafico da apresentação 
from pptx.chart.data import CategoryChartData

# Classe do módulo chart da biblioteca pptx que tem como objetivo
# definir o tipo do gráfico que será utilizado
from pptx.enum.chart import XL_CHART_TYPE

# Editar ppt

# Criando um slide em branco
slide = apresentacao.slides.add_slide(apresentacao.slide_layouts[6]) 

# Lista de produtos do gráfico
produtos = ["Iphone", "Ipad", "Airpod"] 

# Numero de vendas de cada produto
vendas = [1500, 1000, 2000] 

# column_clustered: Atributo que contém o grafico de colunas
tipo_grafico = XL_CHART_TYPE.COLUMN_CLUSTERED

# Distância horizontal em polegadas do gráfico
x = Inches(1)

# Distância vertical em polegadas do gráfico
y = Inches(1)

# Largura do gráfico em polegadas
largura = Inches(5)

# Altura do gráfico em polegadas
altura = Inches(3)
 
# Ira instanciar a classe CategoryChartData que conterá os dados.
dados_graficos = CategoryChartData()

# categories: Atributo da classe CategoryChartData que recebe os rótulos
# dos dados
dados_graficos.categories = produtos

# add_serie: Método da classe CategoryChartData que insere dados no gráfico
# da apresentação. O método recebe como argumento um nome para a legenda
# dos dados e a lista de valores que serão utilizados (lista de valores
# presentes na variável vendas definida anteriormente).
dados_graficos.add_series("Vendas", vendas)

# add_chart: método do atributo shapes que cria gráficos nas apresentações,
# o método possui como argumentos o tipo do gráfico (tipo_grafico), a posição
# do gráfico (x e y), a largura (largura), a altura(altura) e os dados
# do gráfico (dados_graficos)
slide.shapes.add_chart(tipo_grafico, x, y, largura, altura, dados_graficos)

# Salvar ppt
apresentacao.save("MeuPPT3.pptx")
```
