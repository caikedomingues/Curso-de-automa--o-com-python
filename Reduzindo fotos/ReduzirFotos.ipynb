```python
# Módulo da biblioteca pillow que possui métodos e atributos que
# manipulam imagens
from PIL import Image 

# Biblioteca que possibilita que o python acesse e manipule os diretórios
# do sistema de arquivos do sistema operacional.
import os
```

Reduzindo as imagens

```python
# Ira conter o caminho da pasta que contém as imagens que serão reduzidas 
pasta = "imagens"

# Ira conter o caminho da pasta que conterá as imagens reduzidas.
pasta_reduzida = "imagens reduzidas"

# Ira listar os arquivos e diretórios da pasta de imagens
lista_arquivos = os.listdir(pasta)

# Ira percorrer a pasta de imagens com o objetivo de acessar 
# cada arquivo presente na pasta
for arquivo in lista_arquivos:
    
    # open: Função do módulo image que tem como objetivo acessar arquivos. A função
    # recebe como argumento o caminho do arquivo que será acessado/aberto.
    # Ela retorna um objeto que possui os atributos e métodos de manipulação
    # de imagens
    # join: Função do módulo path da biblioteca os que tem como objetivo
    # concatenar caminhos de arquivos e diretórios. A função irá juntar
    # o nome da pasta e o nome do arquivo acessado.
    imagem = Image.open(os.path.join(pasta, arquivo))
    
    # Ira juntar os nomes da pasta de imagens reduzidas e o arquivo
    # que será otimizado.
    novo_nome = os.path.join(pasta_reduzida, "reduzido_" + arquivo)
    
    # Método da classe retornada pela função open que ira salvar
    # as alterações. O método recebe como argumento o caminho do local 
    # que o arquivo irá ser salvo.
    # Optimize: argumento do método save que tem como objetivo diminuir
    # o tamanho da imagem
    imagem.save(novo_nome, optimize= True)
```

Utilizando o convert na diminuição de arquivos

```python
for arquivo in lista_arquivos:
    
    imagem2 = Image.open(os.path.join(pasta, arquivo))
    
    # P: O modo Paletizado limita a imagem a um máximo de 256 cores. Em vez de carregar milhões de cores na memória, o arquivo passa a carregar apenas uma tabela (uma paleta) com 256 cores, e cada pixel só precisa apontar para o número da cor nessa tabela. Isso diminui o tamanho do arquivo quase pela metade.
    
    # Esse algoritmo analisa a nossa imagem, específica e calcula: "Quais são as 256 cores mais importantes e mais presentes aqui?"
    imagem_convertida = imagem2.convert("P", palette=Image.ADAPTIVE)
    
    novo_nome2 = os.path.join(pasta_reduzida, "convertida_" + arquivo)
    
    imagem2.save(novo_nome2)
```
