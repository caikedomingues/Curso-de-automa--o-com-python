```python
# Biblioteca que possibilita que o python se conecte com o sistema de arquivos
# do seu sistema operacional.
import os
```

```python
# Ira conter o path (caminho) da pasta de downloads 
caminho = "C:/Users/caike/Downloads"

# Função da biblioteca os que tem como objetivo listar os arquivos e pastas
# de um diretório. A função recebe como argumento o caminho da pasta que terá
# os seus arquivos listados  
lista_arquivos = os.listdir(caminho)

# Impressão da lista de arquivos
lista_arquivos
```

```python
# Lista que irá conter as datas e os nomes dos arquivos da pasta de downloads
lista_datas = []

# Ira percorrer a lista de arquivos com o objetivo de acessar todos os itens
# presentes na pasta
for arquivo in lista_arquivos:
        
        # getatime: Função que retorna a quantidade de segundos que
        # que se passaram desde o dia 1 de janeiro de 1970 até o exato.
        # momento da ultima ação no arquivo. A função ira receber como
        # parâmetro os caminhos de cada arquivo da pasta download.
        data = os.path.getatime(f"{caminho}/{arquivo}")
        
        # Ira inserir as "datas" e o nome dos arquivos na lista.
        # Observação: Como posteriormente usaremos a função sort para ordenar os arquivos, é necessário passar a data antes do nome do arquivo. Essa
        # ação é necessária, pois, o sort sempre ordena os valores pelo tipo
        # (texto ou numérico) do primeiro valor passado.      
        lista_datas.append((data, arquivo))
```

```python
# A função sort tem como objetivo ordenar os dados de um conjunto de valores.
# Como vimos anteriormente, ela sempre considera o tipo de dado do primeiro
# valor da lista (no nosso caso, as datas). Ela recebe como argumento um reverse
# que possui um valor booleano:
# True: para ordenações em ordem decrescente
# false (valor padrão da função): para ordenações em ordem crescente
lista_datas.sort(reverse=True)

# Ira mostrar os valores da lista
lista_datas
```

```python
# Ira acessar o primeiro arquivo que representa o item mais recente da lista
ultimo_arquivo = lista_datas[0]

ultimo_arquivo
```
